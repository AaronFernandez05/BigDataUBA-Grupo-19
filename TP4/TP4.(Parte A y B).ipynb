```python
# Parte A para 2024.
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
respondieron_2024 = pd.read_csv("C:/Users/CRIS/Documents/Big Data/Big Data Grupo 19/TP2/respondieron_2024.csv")
```

```python
# Variable dependiente
respondieron_2024["desocupado"] = (respondieron_2024["CAT_OCUP"] == 2).astype(int)
```

```python
#Selección y renombrado de variables
variables_2024 = respondieron_2024[["CH04", "CH06", "NIVEL_ED", "REGION", "PONDERA", "desocupado"]].dropna()
variables_2024 = variables_2024.rename( columns={
    "CH04": "sexo",
    "CH06": "edad",
    "NIVEL_ED": "nivel_educativo",
    "REGION": "region",
    "PONDERA": "pondera",
   
})
```

```python
# Convertir variables categóricas en dummies (ANTES de dividir)
df2024_dummies = pd.get_dummies(variables_2024, columns=["sexo", "nivel_educativo", "region"], drop_first=True)
```

```python
# Separar variables dependientes e independientes
X = df2024_dummies.drop(columns="desocupado")
y = df2024_dummies["desocupado"]
```

```python
# Agregar constante 
X["constante"] = 1
```

```python
# División train/test 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=444)
```

```python
# Calcular diferencia de medias
mean_train = X_train.mean(numeric_only=True)
mean_test = X_test.mean(numeric_only=True)
diff_means = (mean_train - mean_test).to_frame(name="Diferencia de medias")
```

```python
# Mostrar diferencias
print(diff_means)
```

```python
# Reducimos decimlaes para que sea mas legible
diff_means = diff_means.round(3)
print(diff_means)
```

```python
#Parte B
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt

```

```python
#Volvemos a darle nombre a las variables
respondieron_2024 = respondieron_2024.rename( columns={
    "CH04": "sexo",
    "CH06": "edad",
    "NIVEL_ED": "nivel_educativo",
    "REGION": "region",
    "PONDERA": "pondera",
   
})
respondieron_2004 = respondieron_2004.rename(columns={
    "CH04": "sexo",
    "CH06": "edad",
    "NIVEL_ED": "nivel_educativo",
    "REGION": "region",
    "PONDERA": "pondera",
    "DESOCUPADO": "desocupado",
    

})
```

```python
# Cargamos el DataFrame limpio 
respondieron_2024 = respondieron_2024[["sexo", "edad", "nivel_educativo", "region", "pondera", "P47T", "PP3E_TOT", "PP3F_TOT"]]
respondieron_2004 = respondieron_2004[["sexo", "edad", "nivel_educativo", "region", "pondera", "P47T", "PP3E_TOT", "PP3F_TOT"]]
```

```python
# Convertimos columnas si es necesario
respondieron_2024["PP3F_TOT"] = pd.to_numeric(respondieron_2024["PP3F_TOT"], errors="coerce")
respondieron_2024["PP3E_TOT"] = pd.to_numeric(respondieron_2024["PP3E_TOT"], errors="coerce")
respondieron_2024["P47T"] = pd.to_numeric(respondieron_2024["P47T"], errors="coerce")
respondieron_2024["edad"] = pd.to_numeric(respondieron_2024["edad"], errors="coerce")
respondieron_2024["sexo"] = pd.to_numeric(respondieron_2024["sexo"], errors="coerce")
respondieron_2024["nivel_educativo"] = respondieron_2024["nivel_educativo"].astype("category")
respondieron_2024["region"] = respondieron_2024["region"].astype("category")
```

```python
# Crear variables necesarias
respondieron_2024 = respondieron_2024.copy()
respondieron_2024["salario_semanal"] = respondieron_2024["P47T"] / 4  # Aproxima salario semanal
respondieron_2024["edad"] = respondieron_2024["edad"].astype(int)
respondieron_2024["edad2"] = respondieron_2024["edad"] ** 2
respondieron_2024["mujer"] = (respondieron_2024["sexo"] == 2).astype(int)
respondieron_2024["HORASTRAB"] = respondieron_2024["PP3E_TOT"].fillna(0) + respondieron_2024["PP3F_TOT"].fillna(0)
```

```python
# Crear dummies de nivel educativo y region
educ_dummies = pd.get_dummies(respondieron_2024["nivel_educativo"], prefix="educ", drop_first=True)
region_dummies = pd.get_dummies(respondieron_2024["region"], prefix="region", drop_first=True)
```

```python
# Unir dummies al DataFrame
model = respondieron_2024(respondieron_2024["salario_semanal", "edad", "edad2", "mujer", "HORASTRAB", educ_dummies, region_dummies], axis=1)
```

```python
# Separar entrenamiento y testeo
train_df, test_df = train_test_split(model, test_size=0.3, random_state=42)
```

```python
# Función para ajustar modelo y devolver resumen
def ajustar_modelo(vars_x, train_df):
    X = sm.add_constant(train_df[vars_x])
    Y = train_df["salario_semanal"]
    model = sm.OLS(Y, X).fit()
    return model
```

```python
# Definir las variables de cada modelo
variables_modelo = {
    "Modelo 1": ["edad"],
    "Modelo 2": ["edad", "edad2"],
    "Modelo 3": ["edad", "edad2"] + list(educ_dummies.columns),
    "Modelo 4": ["edad", "edad2"] + list(educ_dummies.columns) + ["mujer"],
    "Modelo 5": ["edad", "edad2"] + list(educ_dummies.columns) + ["mujer", "HORASTRAB"] + list(region_dummies.columns),
}
```

```python
# Ajustar los modelos
modelos = {nombre: ajustar_modelo(vars_x, train_df) for nombre, vars_x in variables_modelo.items()}
modelos_resumen = {k: v.summary() for k, v in modelos.items()}
```
