```python
# Parte A para 2024.
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("C:/Users/Administrator/Downloads/Big Data/TPs/TP2/respondieron_2024.csv")
```

```python
# Variable dependiente
df["desocupado"] = (df["CAT_OCUP"] == 2).astype(int)
```

```python
#Selección y renombrado de variables
df = df[["CH04", "CH06", "NIVEL_ED", "REGION", "PONDERA", "desocupado"]].dropna()
df = df.rename( columns={
    "CH04": "sexo",
    "CH06": "edad",
    "NIVEL_ED": "nivel_educativo",
    "REGION": "region",
    "PONDERA": "pondera"
})
```

```python
# Convertir variables categóricas en dummies (ANTES de dividir)
df_dummies = pd.get_dummies(df, columns=["sexo", "nivel_educativo", "region"], drop_first=True)
```

```python
# Separar variables dependientes e independientes
X = df_dummies.drop(columns="desocupado")
y = df_dummies["desocupado"]
```

```python
# Agregar constante 
X["constante"] = 1
```

```python
# División train/test 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=444)
```

```python
# 
print(X_train.dtypes)
```

```python
# Calcular diferencia de medias
mean_train = X_train.mean(numeric_only=True)
mean_test = X_test.mean(numeric_only=True)
diff_means = (mean_train - mean_test).to_frame(name="Diferencia de medias")
```

```python
# Mostrar diferencias
print(diff_means)
```

```python
# Reducimos decimlaes para que sea mas legible
diff_means = diff_means.round(3)
print(diff_means)
```
