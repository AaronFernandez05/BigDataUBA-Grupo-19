```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error, mean_absolute_error

# VARIABLE DEPENDIENTE
y_train = train_data["salario_semanal"]
y_test = test_data["salario_semanal"]

# ------------------------
# 1. MODELOS DE ENTRENAMIENTO
# ------------------------

X1 = train_data[["edad"]]
X2 = train_data[["edad", "edad2"]]
X3 = train_data[["edad", "edad2", "educ"]]  # educ numérica
X4 = train_data[["edad", "edad2", "educ", "mujer"]]

# Modelo 5: incluye dummies de región
region_dummies_train = pd.get_dummies(train_data["region"], prefix="region", drop_first=True)
X5 = pd.concat([
    train_data[["edad", "edad2", "educ", "mujer"]],
    region_dummies_train
], axis=1)

X_list = [X1, X2, X3, X4, X5]

# ------------------------
# 2. ENTRENAR LOS MODELOS
# ------------------------

def fit_model(X, y):
    X_const = sm.add_constant(X)
    return sm.OLS(y, X_const).fit()

models = [fit_model(X, y_train) for X in X_list]

# ------------------------
# 3. PREPARAR DATOS DE TEST
# ------------------------

X_test_1 = test_data[["edad"]]
X_test_2 = test_data[["edad", "edad2"]]
X_test_3 = test_data[["edad", "edad2", "educ"]]  # educ numérica
X_test_4 = test_data[["edad", "edad2", "educ", "mujer"]]

# Modelo 5: preparar dummies de región igual que en entrenamiento
region_dummies_test = pd.get_dummies(test_data["region"], prefix="region", drop_first=True)
region_dummies_test = region_dummies_test.reindex(columns=region_dummies_train.columns, fill_value=0)

X_test_5 = pd.concat([
    test_data[["edad", "edad2", "educ", "mujer"]],
    region_dummies_test
], axis=1)

X_test_list = [X_test_1, X_test_2, X_test_3, X_test_4, X_test_5]

# ------------------------
# 4. EVALUACIÓN EN TEST
# ------------------------

metricas = {}

for i, model in enumerate(models):
    X_test_i = sm.add_constant(X_test_list[i], has_constant='add')
    X_test_i = X_test_i[model.model.exog_names]  # alinear columnas con entrenamiento

    y_pred = model.predict(X_test_i)

    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)

    metricas[f"Modelo {i+1}"] = {
        "MSE test": round(mse, 2),
        "RMSE test": round(rmse, 2),
        "MAE test": round(mae, 2)
    }

# ------------------------
# 5. MOSTRAR TABLA DE MÉTRICAS
# ------------------------

tabla_metricas_df = pd.DataFrame(metricas).T
display(tabla_metricas_df)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score, accuracy_score
import matplotlib.pyplot as plt

# ------------------------
# 1. VARIABLE DEPENDIENTE BINARIA
# ------------------------

# Clasificación: si tiene ingreso positivo
y_clas_train = (train_data["salario_semanal"] > 0).astype(int)
y_clas_test = (test_data["salario_semanal"] > 0).astype(int)

# Variables explicativas base
X_clas_train = train_data[["edad", "edad2", "educ", "mujer"]]
X_clas_test = test_data[["edad", "edad2", "educ", "mujer"]]

# ------------------------
# 2. REGRESIÓN LOGÍSTICA
# ------------------------

logit_model = LogisticRegression()
logit_model.fit(X_clas_train, y_clas_train)

# Predicciones de probabilidad y clases
y_pred_proba_logit = logit_model.predict_proba(X_clas_test)[:, 1]
y_pred_logit = (y_pred_proba_logit > 0.5).astype(int)

# ------------------------
# 3. KNN (K=5)
# ------------------------

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_clas_train, y_clas_train)

y_pred_proba_knn = knn_model.predict_proba(X_clas_test)[:, 1]
y_pred_knn = (y_pred_proba_knn > 0.5).astype(int)

# ------------------------
# 4. MÉTRICAS DE EVALUACIÓN
# ------------------------

def evaluar_modelo(y_test, y_pred, y_proba, nombre_modelo):
    acc = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)
    cm = confusion_matrix(y_test, y_pred)

    print(f"\n--- {nombre_modelo} ---")
    print("Confusion Matrix:\n", cm)
    print(f"Accuracy: {acc:.4f}")
    print(f"AUC: {auc:.4f}")

    # ROC curve
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    plt.plot(fpr, tpr, label=f"{nombre_modelo} (AUC = {auc:.2f})")

# Evaluar Logit
evaluar_modelo(y_clas_test, y_pred_logit, y_pred_proba_logit, "Regresión Logística")

# Evaluar KNN
evaluar_modelo(y_clas_test, y_pred_knn, y_pred_proba_knn, "KNN (K=5)")

# ------------------------
# 5. CURVA ROC
# ------------------------

plt.plot([0, 1], [0, 1], 'k--')  # diagonal
plt.xlabel("FPR (1 - Specificidad)")
plt.ylabel("TPR (Sensibilidad)")
plt.title("Curva ROC")
plt.legend()
plt.grid(True)
plt.show()
```
